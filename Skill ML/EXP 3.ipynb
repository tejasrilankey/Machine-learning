```python
import pandas as pd
import matplotlib.pyplot as plt

# Load Dataset
df = pd.read_csv("/content/titanic.csv")

# Display First 5 Rows
print("FIRST 5 ROWS")
print(df.head())

# Dataset Shape
print("\nDATASET SHAPE")
print(df.shape)

# Column Names
print("\nCOLUMN NAMES")
print(df.columns)

# Data Types
print("\nDATA TYPES")
print(df.dtypes)

# Dataset Information
print("\nDATASET INFORMATION")
df.info()

# Statistical Summary
print("\nSTATISTICAL SUMMARY")
print(df.describe())

# Missing Values
print("\nMISSING VALUES")
print(df.isnull().sum())

# Duplicate Rows
print("\nDUPLICATE ROWS")
print(df.duplicated().sum())

# Survival Count
print("\nPASSENGERS SURVIVED")
print(df["survived"].value_counts())

# Survival Percentage
print("\nPERCENTAGE OF PASSENGERS SURVIVED")
print(df["survived"].value_counts(normalize=True) * 100)

# -----------------------------
# BAR CHART
# -----------------------------
survival_counts = df["survived"].value_counts()

plt.figure(figsize=(6,5))
survival_counts.plot(
    kind="bar",
    color=["red", "green"]
)

plt.title("Passenger Survival Count")
plt.xlabel("Survival Status")
plt.ylabel("Number of Passengers")
plt.xticks([0, 1], ["Not Survived", "Survived"], rotation=0)
plt.grid(axis="y")
plt.show()

# -----------------------------
# PIE CHART
# -----------------------------
plt.figure(figsize=(6,6))
survival_counts.plot(
    kind="pie",
    autopct="%1.1f%%",
    labels=["Not Survived", "Survived"],
    colors=["red", "green"],
    startangle=90
)

plt.title("Passenger Survival Percentage")
plt.ylabel("")
plt.show()
```
